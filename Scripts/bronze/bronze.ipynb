#Read CSV files & Write table

```python
for item in INGESTION_CONFIG:
  print(f"Ingesting {item['path']} -> {item['table']}")

  df = (spark.read.option("header", "true")
    .option("inferSchema", "true")
    .csv(item['path']))
  
  (
    df.write
      .mode("overwrite")
      .format("delta")
      .saveAsTable(f"bronze.{item['table']}")

  )
```
