```python
import pyspark.sql.functions as F
from pyspark.sql.functions import col, trim, StringType
```

#Reading From Silver Table

```python
query = """
SELECT
  S.store_key,
  DFT.store_id,
  D.date_key,
  DFT.date,
  PC.promo_key,
  DFT.foot_traffic
FROM silver.daily_foot_trafic DFT
 LEFT JOIN gold.dim_store S
 ON DFT.store_id = S.store_id 
 LEFT JOIN gold.dim_date D
 ON DFT.date = D.full_day  
 LEFT JOIN gold.dim_promo_calander PC
 ON DFT.date BETWEEN PC.start_date AND PC.end_date
 AND DFT.store_id = PC.store_id

"""

df = spark.sql(query)
```

```python
df.display()
```

#Write into gold table

```python
(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("gold.fact_daily_foot_traffic")
)
```
