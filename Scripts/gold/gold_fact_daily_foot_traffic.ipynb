```python
from pyspark.sql import Window
from pyspark.sql.functions import row_number
```

#Reading From Silver Table

##Create surrogate key

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY date, store_id) AS foot_traffic_key,
    *
FROM silver.daily_foot_trafic
"""

df = spark.sql(query)
```

```python
df.show()
```

#Write into gold table

```python
(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("gold.fact_daily_foot_traffic")
)
```
