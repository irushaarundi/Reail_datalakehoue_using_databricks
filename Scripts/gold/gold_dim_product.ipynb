```python
from pyspark.sql import Window
from pyspark.sql.functions import row_number
```

#Reading From Silver Table

##Create surrogate key

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY product_id, category) AS product_key,
    *
FROM silver.product
"""


df = spark.sql(query)
```

#Write into gold table

```python
(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("gold.dim_product")
)
```

```python
%sql 
SELECT * FROM gold.dim_product
```
