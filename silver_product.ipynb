```python
import pyspark.sql.functions as F
from pyspark.sql.functions import col, trim, StringType
from itertools import chain
from pyspark.sql import Window
```

# Reading From Bronze Table

```python
df = spark.table("workspace.bronze.product")
```

```python
print("rows:", df.count())
print("columns:", len(df.columns))
```

#Data Transform

##Trimming

```python
for field in df.schema.fields:
  if isinstance(field.dataType, StringType):
      df = df.withColumn(field.name,trim(col(field.name)))
```

```python
print("rows:", df.count())
print("columns:", len(df.columns))
```

##Standardize the product name

```python
df = df.withColumn(
    "product_name",
    F.when(F.col("product_id") == "P010", "Rain Jacket - Standerd")
     .when(F.col("product_id") == "P001", "Rain Jacket - Primium")
     .otherwise(F.col("product_name").cast("string"))
)
```

```python
df.select("product_id","product_name").distinct().orderBy("product_name").show(1000, truncate=False )
```

```python
print("rows:", df.count())
print("columns:", len(df.columns))
```

##De-duplicate

```python
dups = (df.groupBy(df.columns)
          .count()
          .filter("count > 1")
          .orderBy(F.desc("count")))

dups.show(1000, truncate=False)
print("duplicated groups:", dups.count())
```

```python
w = Window.partitionBy(*df.columns).orderBy(F.lit(1))
df = df.withColumn("row_count", F.row_number().over(w))
df = df.filter("row_count = 1").drop("row_count") 
```

```python
print("rows:", df.count())
print("columns:", len(df.columns))
```

#Write into silver table

```python
(
    df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("silver.product")
)
```
